```python
conn = sqlite3.connect('customer_churn.db')

sql_query = """ 
        SELECT name
        FROM sqlite_master
        WHERE type = 'table';
"""
tables = pd.read_sql(sql_query, conn)

# Create dataframe for each table 
for table_name in tables['name']:
    df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
    globals()[f"df_{table_name}"] = df
    print(f"Created DataFrame: df_{table_name}")

conn.close()
```

```python
churn_rate = df['churn_flag'].mean()*100
print("Churn Rate:", round(churn_rate,2), '%')
```

```python
#3. Churn by Plan type
churn_plan = (df.groupby('plan_type')['churn_flag'].mean().mul(100).round(2).reset_index(name= 'churn_rate_ptc'))
print(churn_plan)
```

```python
#4                                  (a) Churn by state + sum(revenue) & count of users
#                                   (b) chrun by subscription type + sum(revenue) & count of users
```

```python
# (a) Churn by state + sum(revenue) & count of users
churn_by_state = (df.groupby('state')
    .agg(churn_rate=('churn_flag', 'mean'), revenue=('monthly_charges', 'sum'), users=('customerid', 'count')).reset_index())
print(churn_by_state)
```

```python
# (b) chrun by subscription type + sum(revenue) & count of users
churn_by_subscription = (df.groupby('subscription_type')
    .agg(churn_rate=('churn_flag', 'mean'), revenue=('monthly_charges', 'sum'), users=('customerid', 'count')).reset_index())
print(churn_by_subscription)
```

```python
#5. ARPU-- Average Revenue per user


arpu = df['monthly_charges'].mean()
print("Average Revenue per user:", round(arpu, 2))
```

```python
#6.                               (a) count the days of users has used our service : cancellation date else current date
#                                 (b) Avg customer Tenure
```

```python
# (a) count the days of users has used our service : cancellation date else current date
today = pd.Timestamp.today()

df['Tenure'] = np.where(df['cancellation_date'].notna(), (df['cancellation_date'] - df['subscription_start_date']).dt.days,
                        (today - df['subscription_start_date']).dt.days)
df.head()
```

```python
# (b) Avg customer tenure
avg_tenure = df['Tenure'].mean()
print("Average Tenure Days:", avg_tenure)
```

```python
#7. Revenue at risk - revenue lost by churned users
revenue_at_risk = df.loc[df['churn_flag'] == 1, 'monthly_charges'].sum()
print("revenue at risk in (Rs. K):", revenue_at_risk)
```

```python
#8. Escalation rate
# df['escalations'].unique()

complaints = df[df['complaint_date'].notna()]
escalation_rate = (complaints['escalations'] == 'Y').mean() * 100
print("Escalation rate (% of complaints):", round(escalation_rate, 2), "%")
```

```python
df['complaint_counts'] = df['complaint_date'].notna().astype(int)
```

```python
#9. Avg complaint User
avg_complaints = df['complaint_counts'].sum()/df['customerid'].nunique()
print("Avg complaints by users:", round(avg_complaints, 2))
```

```python
#10. Escalations vs Churn
df['escalations'] = np.where(df['escalations'] == 'Y', 1, 0) #encoding string into int type

corr_df = df[['escalations', 'churn_flag']].dropna()

Correlation = corr_df['escalations'].corr(df['churn_flag'])
print("Correlation between escalations vs. churn_flag is :", round(Correlation, 2)*100, "%")
```

```python
#11. Churn risk
conditions = [
    (df['churn_score'] < 50),
    (df['churn_score'] >= 50) & (df['churn_score'] < 70),
    (df['churn_score'] >= 70)
]

choices = ['low', 'medium', 'high']

df['churn_risk'] = np.select(conditions, choices, default='unknown')
```

```python
# (a) Monthly churn trend (Time series KPI)
df_visual['Cancellation_Month'] = df_visual['cancellation_date'].dt.to_period('M')

churn_trend = df_visual[df_visual['churn_flag'] ==1].groupby('Cancellation_Month').size()

plt.figure(figsize=(8,3))
plt.plot(churn_trend.index.astype(str), churn_trend.values, color='green', marker='o', linestyle='dashed', linewidth=2, markersize=12)
plt.title("Monthly Churn Trend")
plt.xlabel("Month")
plt.ylabel("Churned Customers")
plt.show()
```
